#### Exercício 1

Escreva uma função que receba os valores de três lados de um triângulo e o classifique como "Equilátero" (todos os lados iguais), "Isósceles" (dois lados iguais) ou "Escaleno" (todos os lados diferentes).

```python
# Solução
def triangulo(a,b,c):
    if a == b and b == c:
        return "Equilátero"
    elif a != b and b != c and c != a:
        return "Escaleno"
    else:
        return "Isósceles"
```

```python
# Isosceles
print(triangulo(3,2,3))
print(triangulo(3,3,2))
print(triangulo(2,3,3))
```

```python
# Escaleno
print(triangulo(1,3,2))

# Equilátero
print(triangulo(5,5,5))
```

#### Exercício 7

Crie uma função que receba uma lista de números inteiros e retorne um dicionário contendo a contagem de quantos números são pares e quantos são ímpares.

```python
numeros = [
    2, 8, 7, 42, 89, 15, 63,
    28, 94, 51, 3, 76,
    34, 58, 21, 97, 46,
    12, 68, 84, 25, 9,
    72, 39, 55, 18, 91
]
```

```python
# Solução
def contagem_par_impar(lista):
    par = 0
    impar = 0
    for i in lista:
        if i % 2 == 0:
            par +=1
        else:
            impar += 1
    return {"par": par, "impar": impar}
```

```python
contagem_par_impar(numeros)
```

#### Exercício 8

Crie uma função que receba uma lista de strings (potenciais e-mails) e um parâmetro opcional dominio_desejado (com valor padrão "gmail.com"). A função deve retornar uma nova lista, usando list comprehension, contendo apenas os e-mails que terminam com o domínio desejado.

```python
emails = [
    "ana.clara@gmail.com",
    "lucas.almeida@yahoo.com",
    "mariana.costa@outlook.com",
    "rafael.pereira@hotmail.com",
    "juliana.ribeiro@gmail.com",
    "bruno.martins@yahoo.com",
    "camila.azevedo@outlook.com",
    "diego.nogueira@hotmail.com",
    "fernanda.lima@gmail.com",
    "pedro.henrique@yahoo.com",
    "beatriz.rocha@outlook.com",
    "thiago.moreira@hotmail.com",
    "larissa.pimentel@gmail.com",
    "guilherme.farias@yahoo.com",
    "renata.barros@outlook.com"
]
```

```python
# Solução
# assim analisará apenas se tem o dominio, logo, se algum email começa com o dominio e terminar com outro, vai ser apresentado da mesma forma
def busca_email(lista, dominio_desejado="gmail.com"):
    return [email for email in lista if dominio_desejado in email]
```

```python
busca_email(emails)
```

```python
# Solução
# assim analisará apenas se tem o dominio, logo, se algum email começa com o dominio e terminar com outro, vai ser apresentado da mesma forma
def busca_email2(lista, dominio_desejado="gmail.com"):
    return [email for email in lista if dominio_desejado == email[-len(dominio_desejado):] ]
```

```python
busca_email2(emails)
```

```python
busca_email2(emails,"yahoo.com")
```

Dada uma lista de frases, use a função map() em conjunto com uma expressão lambda para criar uma nova lista onde cada frase é convertida para letras maiúsculas e tem a palavra "PYTHON" anexada ao final.

```python
frases = [
    "O sol brilha mais forte nas manhãs de domingo.",
    "Aprender algo novo todos os dias é essencial para crescer.",
    "A vida é feita de pequenas escolhas que definem grandes caminhos.",
    "Nunca subestime o poder de um sorriso sincero.",
    "A música certa pode transformar qualquer dia cinza em colorido.",
    "Viajar é a melhor forma de colecionar lembranças inesquecíveis.",
    "Um bom livro é um portal para mundos desconhecidos.",
    "A paciência é a chave para superar desafios inesperados.",
    "A amizade verdadeira resiste ao tempo e à distância.",
    "Cada erro é uma oportunidade disfarçada de aprendizado."
]
```

#### Exercício 10

Crie um jogo onde o computador escolhe um número secreto entre 1 e 20. O jogador tem 5 tentativas para adivinhar. A cada tentativa, o programa informa se o palpite foi muito alto ou muito baixo. Se o jogador acertar, o loop deve ser interrompido com uma mensagem de vitória.

```python
# Solução
import random

tentativas = 0
cpu_random = random.randint(1,20)

while tentativas <= 5:
    if tentativas == 5:
        print("Perdeu")
        break
    else:
        guess = int(input("Tente adivinhar qual numero eu escolhi!"))
        if guess > cpu_random:
            print("muito alto")
        elif guess < cpu_random:
            print("muito baixo")
        else:
            print("Vitória")
            break
        tentativas += 1
    
```
