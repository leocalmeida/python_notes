# Estrutura de dados

## Listas (list)

```python
minha_lista = [1, "Leonardo", 3.14, True]
```

```python
print(minha_lista[0])
```

```python
minha_lista.append("Novo item")
```

```python
print(minha_lista)
```

## Tuplas (tuple)

```python
minha_tupla = (1,2,3,4)
```

```python
print(minha_tupla)
```

```python
minha_tupla.append(5) # gera um erro
```

## Conjuntos (set)

```python
meu_set = {1,2,3,4}
```

```python
print(meu_set) # como a ordem nao é garantida, a apresentação pode variar
```

```python
set_a = {1,2,3}
set_b = {2,3,5}
```

```python
set_c = set_a.union(set_b)
```

```python
print(set_c)
```

```python
print(set_c[1]) # vai dar erro
```

```python
print(meu_dict["nome"])
```

```python
print(meu_dict)
```

## Funções

### Função sem parâmetro

```python
def atim():
    print("Saúde!")
```

```python
atim()
```

### Função com parâmetro

```python
def falar_oi(nome):
    print(f"Oi {nome}")
```

```python
falar_oi("leonardo")
```

```python
# definindo um valor padrão caso nenhum dado seja passado
def falar_oi(nome="cleitin"):
    print(f"Oi {nome}")
```

```python
falar_oi()
```

### Função com retorno

```python
def soma(a , b):
    return a + b
```

```python
resultado = soma(1,2)
print(resultado)
```
